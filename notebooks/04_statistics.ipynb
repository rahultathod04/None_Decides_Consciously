```python
# DAY 4 : Statistical Hypothesis Testing
# GOAL  : Prove our findings are REAL with numbers
#         and not just visual coincidence

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
df = pd.read_csv('../data/cleaned/student_lifestyle_cleaned.csv')
```

```python
df.shape
```

```python
#H1

correlation, p_value = stats.pearsonr(df['stress_level'], 
                                       df['daily_social_media_hours'])

print("=" * 45)
print("H1 : Stress ~ Social Media Usage")
print("=" * 45)
print(f"Correlation : {correlation:.4f}")
print(f"P-value     : {p_value:.10f}")
print()
if p_value < 0.05:
    print("PROVED — Relationship is REAL, not luck!")
else:
    print("NOT PROVED — Could be random")
```

```python
#H2
correlation, p_value = stats.pearsonr(df['sleep_hours'],
                                       df['exam_score'])

print("=" * 45)
print("H2 : Sleep Hours → Exam Score")
print("=" * 45)
print(f"Correlation : {correlation:.4f}")
print(f"P-value     : {p_value:.10f}")
print()
if p_value < 0.05:
    print("PROVED — Relationship is REAL, not luck!")
else:
    print("NOT PROVED — Could be random")
```

```python
#H3
correlation, p_value = stats.pearsonr(df['mood_rating'],
                                       df['social_interaction_level'])

print("=" * 45)
print("H3 : Mood Rating → Social Interaction")
print("=" * 45)
print(f"Correlation : {correlation:.4f}")
print(f"P-value     : {p_value:.10f}")
print()
if p_value < 0.05:
    print("PROVED — Relationship is REAL, not luck!")
else:
    print("NOT PROVED — Could be random")
```

```python
# Summary of all 3 hypothesis tests
results = {
    'Hypothesis': [
        'H1: Stress - Social Media',
        'H2: Sleep  - Exam Score',
        'H3: Mood   - Social Interaction'
    ],
    'Correlation': [0.5590, 0.5164, 0.1492],
    'P-value': ['< 0.001', '< 0.001', '< 0.001'],
    'Strength': ['Strong', 'Strong', 'Weak'],
    'Result': ['Proved', 'Proved', 'Weakly Proved']
}
```

```python
summary = pd.DataFrame(results)
print("=" * 65)
print("FINAL STATISTICAL SUMMARY — EmoPredict Project")
print("=" * 65)
print(summary.to_string(index=False))
print("=" * 65)
print("\nConclusion: Emotional STATE drives human BEHAVIOR.")
print("   Not intelligence. Not free will. STATE.")
```
