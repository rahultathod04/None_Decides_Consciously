```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/student_lifestyle.csv")
print(f"✅ Loaded: {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
for col in df.columns:
    print(f"\n {col}")
    print("="*20)
    print(f" type : {df[col].dtype}")
    print(f" null values : {df[col].isnull().sum()}")
    print(f" sample : {df[col].dropna().unique()[:5]}")
    print(f" sample_max : {df[col].dropna().unique().max()}")
    print(f" sample_min : {df[col].dropna().unique().min()}")
```

#Goal
##Total missing values left : 0
##stress_level max          : 10
##study_hours min           : 0.0

```python
# managing stress level(in range of 1-10)
bad_stress = df[df['stress_level'] > 10]
print(bad_stress[['student_id', 'stress_level']])
```

```python
df['stress_level'] = df['stress_level'].clip(upper=10)
print(df["stress_level"].max())
```

```python
# managing the study_per_day > 0
bad_study = df[df['study_hours_per_day'] < 0]
print(bad_study[['student_id', 'study_hours_per_day']])
```

```python
df['study_hours_per_day'] = df['study_hours_per_day'].clip(lower=0)
print(df["study_hours_per_day"].min())
```

```python
# fixing the numaric values
numeric_cols = df.select_dtypes(include='number').columns.tolist()
numeric_cols.remove('student_id')

print(f"Numeric columns: {len(numeric_cols)}")
for col in numeric_cols:
    print(f" {col}")
```

```python
for col in numeric_cols:
    median_val= df[col].median() 
    missing_val = df[col].isnull().sum()
    df[col] = df[col].fillna(median_val)
    print(f"In {col} {missing_val} no. missing values are filled with median of the column {median_val:.2f} ")
```

```python
# fixing the text 
txt_values = df.select_dtypes(include='object').columns.tolist()
print(f"\ntxt_columns: {len(txt_values)}")
for col in txt_values:
    print(f"_{col} with missing valuse {df[col].isnull().sum()}")
```

```python
most_comman= df["platform_usage"].mode()[0]
df['platform_usage'] = df['platform_usage'].fillna(most_comman)
print(f"all missing values of platform usage are filled with {most_comman}")
```

```python
print("=" * 20)
print("CLEANING REPORT")
print("=" * 20)
print(f"Total missing values left : {df.isnull().sum().sum()}")
print(f"stress_level max          : {df['stress_level'].max()}")
print(f"study_hours min           : {df['study_hours_per_day'].min()}")
```

```python
df.to_csv('../data/cleaned/student_lifestyle_cleaned.csv', index=False)
print(" Clean dataset saved to data/cleaned/")
```
